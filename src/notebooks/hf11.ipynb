```
--Task 1: Which 10 companies make the most money?
SELECT 
    dim_company.company,
ROUND(SUM(fact_taxi_trips.trip_total),2) AS trip_total
FROM
    fact_taxi_trips
INNER JOIN
    dim_company 
ON dim_company.company_id=fact_taxi_trips.company_id
GROUP BY
    dim_company.company
ORDER BY
    trip_total DESC
LIMIT 10
```

```
--Task 2: Top 10 pickup community areas with the most rides
SELECT 
    dim_community_areas.community_area,
    count(*) as total_pickup_ride
FROM
    fact_taxi_trips
INNER JOIN
    dim_community_areas
ON dim_community_areas.area_code=fact_taxi_trips.pickup_community_area_id
GROUP BY
    dim_community_areas.community_area
ORDER BY total_pickup_ride DESC
LIMIT 10
```

```
--Task 3: Count taxi rides per day of the week
SELECT
    COUNT(day_of_week) as tripnumber_on_weekday,
CASE
    WHEN DAY_OF_WEEK = 1 THEN 'Monday'
    WHEN DAY_OF_WEEK = 2 THEN 'Tuesday'
    WHEN DAY_OF_WEEK = 3 THEN 'Wednesday'
    WHEN DAY_OF_WEEK = 4 THEN 'Thursday'
    WHEN DAY_OF_WEEK = 5 THEN 'Friday'
    WHEN DAY_OF_WEEK = 6 THEN 'Saturday'
    WHEN DAY_OF_WEEK = 7 THEN 'Sunday'
END AS weekdays
FROM 
    fact_taxi_trips
INNER JOIN 
    dim_date 
ON
    trip_start_timestamp=dim_date.date
GROUP BY day_of_week
ORDER BY day_of_week
```
