```python
from bs4 import BeautifulSoup
import pandas as pd
import requests
```

```python
url="https://en.wikipedia.org/wiki/Community_areas_in_Chicago"
headers={"User-Agent":"Mozilla/5.0"}

response=requests.get(url=url,headers=headers)

response
```

```python
soup=BeautifulSoup(response.content,"html.parser")

soup
```

```python
for table in soup.find_all("table"):
    print (table.get("class"))
```

```python
table=soup.find("table",class_="wikitable sortable plainrowheaders mw-datatable")
table
```

```python
for row in table.tbody.find_all("tr")[2:-1]:    #listából a harmadiktól és az utolsót ne (nem szükséges sorok kiszűrése)
    print (row)
    print ("-" *50)
```

```python
data=[]

for row in table.tbody.find_all("tr")[2:-1]:    #listából a harmadiktól és az utolsót ne (nem szükséges sorok kiszűrése)
    cells= row.find_all("td")
    header_cells= row.find_all("th")
    
    area_code = cells[0].get_text(strip=True)   #ez is egy BSoup parancs, megvannak a körzetszámok
    community_area_name = header_cells[0].get_text(strip=True)   #ez is egy BSoup parancs, megvannak a körzetszámok
    
    data.append({"area_code":area_code,"community name":community_area_name})

 
```

```python
data
```

```python
community_areas_df=pd.DataFrame(data)
community_areas_df["area_code"]=community_areas_df["area_code"].astype("int")
```
